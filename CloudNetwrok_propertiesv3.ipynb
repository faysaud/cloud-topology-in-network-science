```python
import random
import matplotlib.pyplot as plt
# import matplotlib.ticker as mtick
# from matplotlib.ticker import PercentFormatter
import networkx as nx
import pandas as pd
# from os import listdir
# import xlrd
# import warnings
from ipynb.fs.full.Reading_CloudNetworks_f import Networks_DictList

```

```python
def properties(G):
#return properties of graph G if the gaint compponet did not disapear or only contined 1 node 

    Connected_components = list(nx.connected_components(G))
    numConnected_components= len(Connected_components)
    subG = G.subgraph(max(Connected_components, key=len))   
    ASP = nx.average_shortest_path_length(subG)
    diam = nx.diameter(subG)

    degrees_Dict =G.degree(); degrees_list= [v for k, v in degrees_Dict]
    avg_degree= sum(degrees_list)/G.number_of_nodes()
    ACC = nx.average_clustering(G)
    r = nx.degree_assortativity_coefficient(G)
    attR= nx.attribute_assortativity_coefficient(G,'role')   
    numEdges= len(G.edges()) #does not work with edges attacks only for networks proporties 

    return [numConnected_components,len(subG) ,round(ASP,2), round(diam,2),round(ACC,2), round(avg_degree,2), numEdges,r,attR]
```

# Get Network proprties 

```python
networks = Networks_DictList(test=True)
PResults=[];names=[]
for network in networks:
    G = network['graph'].copy()
    PResults.append(properties(G))
    names.append(network['name'])

NetworkDStable = pd.DataFrame()
umConnected_components=[];LCC=[]
diam= [];ASP=[]
ACC=[]; avg_degree=[]
numEdges=[];r=[];attR=[]

for propertie in PResults:
    umConnected_components.append(propertie[0])
    LCC.append(propertie[1])
    ASP.append(propertie[2])
    diam.append(propertie[3])
    ACC.append(propertie[4])
    avg_degree.append(propertie[5])
    numEdges.append(propertie[6])
    r.append(propertie[7])
    attR.append(propertie[8])
```

```python
NetworkDStable["Networks"]=names
NetworkDStable["#CC"]= umConnected_components
NetworkDStable["largest_component size"]=LCC ;NetworkDStable["Diameter"]=diam
NetworkDStable["ASP"] = ASP; NetworkDStable["Average Clustering Coefficient"] =ACC
NetworkDStable["avgDegress"] = avg_degree; NetworkDStable["# Edges"] =numEdges
NetworkDStable[" R"] =r ; NetworkDStable["role R coef."] =attR
NetworkDStable
# NetworkDStable.to_csv("Network datasets properties2")
```

```python
from collections import Counter 

def DegDist (networks):
    for network in networks:
        degrees = [network['graph'].degree(n) for n in network['graph'].nodes()]
        degree_count = Counter(degrees)
        x, y = zip(*sorted(degree_count.items()))

        fig,ax = plt.subplots()
        plt.bar(x, y)

        plt.title(network['name'])

networks = Networks_DictList()
DegDist (networks)
```

```python
from collections import Counter 

# distribution of types 

def rolePer ():
    networksDist=[]
    for network in networks:
        G=network['graph']
        name=network['name']
        attributes= nx.get_node_attributes(G,"role")
        networkSize=G.number_of_nodes()
        print(name,networkSize)
        typesList=[v for k, v in attributes.items()]
        typeCountDict=Counter(typesList)
        types=[k for k, v in typeCountDict.items()]
        countList=[v for k, v in typeCountDict.items()]
        pers=[(v/networkSize)*100 for v in countList ]
        networksDist.append([name,types,pers])

    return networksDist 


networks = Networks_DictList()
RolesTable = pd.DataFrame()
name= [];role=[];per=[]
networksDist=rolePer()

for network in networksDist:
    name.append(network[0])
    role.append(network[1])
    per.append(network[2])


RolesTable['netwrok']=name
RolesTable['role']=role
RolesTable['per']=per

RolesTable
RolesTable.to_csv('rolePer100.csv')
```
