```python
import pandas 
import os
import csv
```

```python
def csv_to_latex(metricIndexc,file,outputFileName=None):
    listOfLists=[]
    ASPVals=[]
    metricV=[0]
    data=pandas.read_csv(file,header=0)
    per=data.iloc[:5,2].tolist()
    # asp=[0]
    with open(file, mode ='r')as file:
        csvFile = csv.reader(file)
        next(csvFile)
        for lines in csvFile:
            metricV.append(lines[metricIndexc])

    for network in range(1,len(metricV)):
        ASPVals.append(metricV[network])
        if network % 5 == 0 :
            listOfLists.append(ASPVals)
            ASPVals=[]   
    finaltabel = pandas.DataFrame(listOfLists, columns = per)
    print(finaltabel)
    return finaltabel.to_latex(outputFileName)
```

```python
# metricIndexc > a int of the metrci index to retune the values of the metric only take int for ASP = 6 ,ACC = 7,avgDeg=8
# name of file to conver to latex
# third paramneter =  take a file name to Store output if not provided the latex will be returned as string 
latex_file= csv_to_latex(6,"Random_nodesAttcak.csv","Random_nodesAttcak_latexASP.txt")
```
